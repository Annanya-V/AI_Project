```python
from langchain_community.document_loaders import PyPDFLoader
def load_pdf_file(file_path):
    loader = PyPDFLoader(file_path)
    documents = loader.load()
    return documents
```

```python
extracted_data = load_pdf_file("Law_Handbook_text.pdf")
```

```python
#Split the Data into Text Chunks
def text_split(extracted_data):
    text_splitter=RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=20)
    text_chunks=text_splitter.split_documents(extracted_data)
    return text_chunks
```

```python
text_chunks=text_split(extracted_data)
print("Length of Text Chunks", len(text_chunks))
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
        model="openai/gpt-oss-20b",
        temperature=0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

system_prompt = (
    "You are an assistant for question-answering tasks. "
    "Use the following pieces of retrieved context to answer the question. "
    "Use the chat history to understand follow-up questions. "
    "If you don't know the answer, say that you don't know. "
    "Use three sentences maximum and keep the answer concise."
    "\n\nContext:\n{context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    MessagesPlaceholder("history"),
    ("human", "{input}"),
])

condense_prompt = ChatPromptTemplate.from_messages([
    ("system",
     "Given the chat history and the user's latest message, rewrite the latest "
     "message as a single standalone question that can be understood without the "
     "history. If it is already standalone, return it unchanged. "
     "Return ONLY the question, with no explanation."),
    MessagesPlaceholder("history"),
    ("human", "{input}"),
])
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

def clean_history(history):
    out = []
    for m in (history or [])[-6:]:
        role = "ai" if m.get("role") in ("assistant", "bot", "ai") else "human"
        out.append((role, str(m.get("content", ""))[:1000]))
    return out

def standalone_question(question, history):
    if not history:
        return question
    msgs = condense_prompt.format_messages(history=history, input=question)
    return llm.invoke(msgs).content.strip() or question

def answer_question(question, history, extra_context=""):
    history = clean_history(history)
    search_q = standalone_question(question, history)
    docs = retriever.invoke(search_q)
    context = format_docs(docs)
    if extra_context:
        context = "User's uploaded document:\n" + extra_context + "\n\nKnowledge base:\n" + context
    msgs = prompt.format_messages(context=context, history=history, input=question)
    return llm.invoke(msgs).content, docs
```

```python
answer, docs = answer_question("what is tort?", [])
print(answer)

# follow-up test
history = [
    {"role": "user", "content": "what is tort?"},
    {"role": "assistant", "content": answer},
]
answer2, docs2 = answer_question("what leads to it?", history)
print(answer2)
```

```python
from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from typing import List
import nest_asyncio, uvicorn, threading
from pyngrok import ngrok

app = FastAPI()

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"],
)

class Query(BaseModel):
    question: str
    history: List[dict] = []

@app.get("/")
def health():
    return {"status": "ok"}

@app.post("/ask")
def ask(q: Query):
    try:
        answer, docs = answer_question(q.question, q.history)
        sources = [
            {"page": d.metadata.get("page"), "snippet": d.page_content[:200]}
            for d in docs
        ]
        return {"answer": answer, "sources": sources}
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))

ngrok.set_auth_token("")
public_url = ngrok.connect(8000)
print("Backend URL:", public_url)

nest_asyncio.apply()
threading.Thread(
    target=lambda: uvicorn.run(app, host="0.0.0.0", port=8000),
    daemon=True,
).start()
```

```python
import io, pytesseract
from PIL import Image
from pypdf import PdfReader
import docx2txt
from fastapi import UploadFile, File, Form

def extract_text(name, data):
    n = name.lower()
    if n.endswith(".pdf"):
        reader = PdfReader(io.BytesIO(data))
        return "\n".join(p.extract_text() or "" for p in reader.pages)
    if n.endswith((".png", ".jpg", ".jpeg", ".webp")):
        return pytesseract.image_to_string(Image.open(io.BytesIO(data)))
    if n.endswith(".docx"):
        return docx2txt.process(io.BytesIO(data))
    if n.endswith(".txt"):
        return data.decode("utf-8", errors="ignore")
    raise ValueError("Unsupported file type")

@app.post("/ask_file")
async def ask_file(question: str = Form(...), file: UploadFile = File(...)):
    try:
        data = await file.read()
        text = extract_text(file.filename, data)[:6000]
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))
    if not text.strip():
        raise HTTPException(status_code=400, detail="No readable text found in the file")

    docs = retriever.invoke(question)
    context = (
        "User's uploaded document:\n" + text +
        "\n\nKnowledge base:\n" + format_docs(docs)
    )
    messages = prompt.format_messages(context=context, input=question)
    answer = llm.invoke(messages).content
    sources = [{"page": d.metadata.get("page"), "snippet": d.page_content[:200]} for d in docs]
    return {"answer": answer, "sources": sources}
```
